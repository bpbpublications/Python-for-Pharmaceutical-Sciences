## 01. Worked Example 6.1 - Display Sample Numbers Using a for Loop

### Problem statement

Display laboratory sample numbers from 1 to 5.

### Objective

Use a for loop to repeat a simple output statement a known number of times.

### Pharmacy context

This pattern can label a small set of simulated laboratory samples, dissolution vessels or replicate observations during classroom exercises.

### Algorithm

1. Start.
2. Generate integers 1 through 5 using range(1, 6).
3. For each value, store it in sample_no.
4. Display the current sample number.
5. Stop after the range is exhausted.

### Code and executed output

The output appears directly below the code cell.

```python
for sample_no in range(1, 6):
    print("Sample number:", sample_no)
```

### Explanation

range(1, 6) generates 1, 2, 3, 4 and 5. The stop value 6 is excluded. The indented print() statement runs once for each generated number.

### Interpretation

This pattern can label a small set of simulated laboratory samples, dissolution vessels or replicate observations during classroom exercises.

## 03. Worked Example 6.3 - Countdown of Samples Remaining

### Problem statement

A classroom activity has five samples left to inspect. Display the number remaining until all five have been processed.

### Objective

Demonstrate initialization, condition testing and update in a while loop.

### Pharmacy context

A countdown loop can represent a controlled repeated workflow, such as completing a fixed queue of simulated samples.

### Algorithm

1. Set samples_remaining to 5.
2. While samples_remaining is greater than 0, display its value.
3. Decrease samples_remaining by 1.
4. When the condition becomes False, display a completion message.

### Code and executed output

The output appears directly below the code cell.

```python
samples_remaining = 5

while samples_remaining > 0:
    print("Samples remaining:", samples_remaining)
    samples_remaining -= 1

print("Inspection sequence complete")
```

### Explanation

The variable decreases by 1 after every iteration. When it reaches 0, the condition samples_remaining > 0 becomes False and the loop ends.

### Interpretation

A countdown loop can represent a controlled repeated workflow, such as completing a fixed queue of simulated samples.

## 04. Worked Example 6.4 - Count Simulated Assay Results Meeting a Classroom Criterion

### Problem statement

Enter five simulated assay percentages and count how many fall within the classroom band 95.0% to 105.0%.

### Objective

Use a counter with a for loop and a conditional statement.

### Pharmacy context

The example illustrates how repeated QC-like observations can be classified and counted. The numerical band is explicitly a classroom value for programming practice.

### Algorithm

1. Set accepted_count to 0.
2. Repeat five times.
3. Read assay_percent.
4. If 95.0 <= assay_percent <= 105.0, increase accepted_count by 1.
5. After the loop, display accepted_count.

### Important notes

Saved output uses these sample entries in order: `99.2`, `94.6`, `101.1`, `105.4`, `98.7`. Set INTERACTIVE_INPUT = True in the setup cell to type your own entries.

### Code and executed output

The output appears directly below the code cell.

```python
_sample_answers = iter(['99.2', '94.6', '101.1', '105.4', '98.7'])

accepted_count = 0

for sample_no in range(1, 6):
    assay_percent = float(classroom_input(f"Enter assay % for sample {sample_no}: "))
    if 95.0 <= assay_percent <= 105.0:
        accepted_count += 1

print("Results inside classroom band:", accepted_count)
```

### Explanation

accepted_count changes only when the condition is True. It is therefore a counter, not a running total of the assay values.

### Interpretation

The example illustrates how repeated QC-like observations can be classified and counted. The numerical band is explicitly a classroom value for programming practice.

## 05. Worked Example 6.5 - Accumulate Tablet Weights and Compute a Mean

### Problem statement

Enter the weights of four simulated tablets and calculate the total and mean weight.

### Objective

Use an accumulator to add repeated numerical values.

### Pharmacy context

Accumulation is a basic pattern used before calculating totals and averages of repeated measurements. The data shown are simulated.

### Algorithm

1. Set total_weight_mg to 0.
2. Repeat four times.
3. Read weight_mg.
4. Add weight_mg to total_weight_mg.
5. Divide total by 4 to obtain the mean.
6. Display total and mean.

### Important notes

Saved output uses these sample entries in order: `501`, `498`, `503`, `500`. Set INTERACTIVE_INPUT = True in the setup cell to type your own entries.

### Code and executed output

The output appears directly below the code cell.

```python
_sample_answers = iter(['501', '498', '503', '500'])

total_weight_mg = 0.0
number_of_tablets = 4

for tablet_no in range(1, number_of_tablets + 1):
    weight_mg = float(classroom_input(f"Enter weight of tablet {tablet_no} (mg): "))
    total_weight_mg += weight_mg

mean_weight_mg = total_weight_mg / number_of_tablets
print("Total weight (mg):", total_weight_mg)
print("Mean weight (mg):", mean_weight_mg)
```

### Explanation

total_weight_mg is an accumulator because every entered weight is added to the running total. The mean is calculated only after all four values have been processed.

### Interpretation

Accumulation is a basic pattern used before calculating totals and averages of repeated measurements. The data shown are simulated.

## 06. Worked Example 6.6 - Batch and Replicate Labels

### Problem statement

Generate labels for 3 simulated batches, each having 2 replicate measurements.

### Objective

Demonstrate two-level repetition using nested for loops.

### Pharmacy context

Nested loops are useful when classroom data have two repeated dimensions, such as batches and replicates, formulations and time points, or rows and columns of a small plate layout.

### Algorithm

1. Repeat batch_no from 1 to 3.
2. For each batch, repeat replicate_no from 1 to 2.
3. Display the batch-replicate combination.
4. Stop after all combinations are produced.

### Code and executed output

The output appears directly below the code cell.

```python
for batch_no in range(1, 4):
    for replicate_no in range(1, 3):
        print("Batch", batch_no, "- Replicate", replicate_no)
```

Nested loops are useful when classroom data have two repeated dimensions, such as batches and replicates, formulations and time points, or rows and columns of a small plate layout.

## 07. Worked Example 6.7 - Stop When a Temperature Excursion Is Detected

### Problem statement

Up to six simulated storage-temperature readings will be entered. Stop processing immediately if a value exceeds 8.0 °C.

### Objective

Use break to terminate a loop when a stop condition is met.

### Pharmacy context

The pattern demonstrates early termination when a classroom-defined alert condition is detected. Actual storage specifications must come from approved product and regulatory requirements.

### Algorithm

1. Repeat for a maximum of six readings.
2. Enter temperature_c.
3. If temperature_c > 8.0, display an excursion message and break.
4. Otherwise display that the reading was processed.
5. Continue until six readings are processed or break occurs.

### Important notes

This specific example checks only an upper threshold. It does not implement a complete lower-and-upper storage-range screen.

Saved output uses these sample entries in order: `5.1`, `6.0`, `8.6`. Set INTERACTIVE_INPUT = True in the setup cell to type your own entries.

### Code and executed output

The output appears directly below the code cell.

```python
_sample_answers = iter(['5.1', '6.0', '8.6'])

for reading_no in range(1, 7):
    temperature_c = float(classroom_input(f"Reading {reading_no} (°C): "))

    if temperature_c > 8.0:
        print("Classroom excursion flag - stopping input")
        break

    print("Reading processed")
```

### Explanation

When the third value meets the stop condition, break exits the for loop. Readings 4 to 6 are not requested.

### Interpretation

The pattern demonstrates early termination when a classroom-defined alert condition is detected. Actual storage specifications must come from approved product and regulatory requirements.

## 08. Worked Example 6.8 - Skip a Missing Measurement

### Problem statement

Enter five simulated pH readings. Use -1 as a classroom code for a missing reading. Skip missing values and add only available readings to the total.

### Objective

Use continue to ignore a selected iteration without terminating the loop.

### Pharmacy context

Missing-data handling should be explicit. This classroom example shows how a program can skip a designated missing marker while preserving the repeated workflow.

### Algorithm

1. Set total_ph and valid_count to 0.
2. Repeat five times.
3. Read ph_value.
4. If ph_value == -1, display a missing message and continue.
5. Otherwise add ph_value to total_ph and increase valid_count.
6. After the loop, display valid_count and total_ph.

### Important notes

The sum of the pH numbers demonstrates an accumulator only; pH is logarithmic, so this sum is not a scientifically meaningful combined pH measurement.

Saved output uses these sample entries in order: `6.8`, `-1`, `7.0`, `6.9`, `7.1`. Set INTERACTIVE_INPUT = True in the setup cell to type your own entries.

### Code and executed output

The output appears directly below the code cell.

```python
_sample_answers = iter(['6.8', '-1', '7.0', '6.9', '7.1'])

total_ph = 0.0
valid_count = 0

for reading_no in range(1, 6):
    ph_value = float(classroom_input(f"pH reading {reading_no} (-1 if missing): "))

    if ph_value == -1:
        print("Missing reading skipped")
        continue

    total_ph += ph_value
    valid_count += 1

print("Valid readings:", valid_count)
print("Total of valid pH readings:", total_ph)
```

### Explanation

The continue statement prevents the missing code -1 from being included in the accumulator or count. The loop itself continues with the next reading.

### Interpretation

Missing-data handling should be explicit. This classroom example shows how a program can skip a designated missing marker while preserving the repeated workflow.

## 09. Worked Example 6.9 - Count ADR Classroom Entries Until END

### Problem statement

Accept simulated ADR category entries until the user types END. Count how many non-empty entries are recorded.

### Objective

Use a while loop controlled by a text sentinel.

### Pharmacy context

The example models open-ended data-entry logic using simulated ADR labels. It is a programming exercise, not a pharmacovigilance reporting system.

### Algorithm

1. Set report_count to 0.
2. Read adr_entry.
3. While adr_entry is not END, process the entry.
4. If the entry is not blank, increase report_count.
5. Read the next entry.
6. When END is entered, leave the loop and display report_count.

### Important notes

Saved output uses these sample entries in order: `nausea`, `rash`, `dizziness`, `END`. Set INTERACTIVE_INPUT = True in the setup cell to type your own entries.

### Code and executed output

The output appears directly below the code cell.

```python
_sample_answers = iter(['nausea', 'rash', 'dizziness', 'END'])

report_count = 0
adr_entry = classroom_input("Enter ADR category or END: ").strip()

while adr_entry.upper() != "END":
    if adr_entry != "":
        report_count += 1
    adr_entry = classroom_input("Enter ADR category or END: ").strip()

print("Classroom entries recorded:", report_count)
```

### Explanation

The loop does not need to know the number of entries beforehand. It stops only when END is entered. strip() removes surrounding spaces and upper() makes END detection case-insensitive.

### Interpretation

The example models open-ended data-entry logic using simulated ADR labels. It is a programming exercise, not a pharmacovigilance reporting system.

## 10. 6.17 Integrated Worked Program - Small QC Summary

### Problem statement

Enter a user-selected number of simulated assay results. Skip impossible classroom entries outside 0 to 200%, count valid results, count results inside the classroom band 95 to 105%, and calculate the mean of valid results.

### Objective

Integrate loop control, validation, counting and accumulation in one beginner-level program.

### Pharmacy context

This program demonstrates the logic of processing repeated laboratory-style observations, separating invalid data from analyzable data, summarizing the valid values and counting values in a simulated classroom band.

### Algorithm

1. Read number_of_samples.
2. Initialize total_assay, valid_count and accepted_count to 0.
3. Repeat for each planned sample.
4. Read assay_percent.
5. If assay_percent is outside 0 to 200, display a message and continue.
6. Add valid assay_percent to total_assay and increase valid_count.
7. If 95 <= assay_percent <= 105, increase accepted_count.
8. After the loop, calculate mean_assay only when valid_count > 0.
9. Display the summary.

### Important notes

Saved output uses these sample entries in order: `5`, `99.2`, `102.0`, `-3`, `94.8`, `101.5`. Set INTERACTIVE_INPUT = True in the setup cell to type your own entries.

### Code and executed output

The output appears directly below the code cell.

```python
_sample_answers = iter(['5', '99.2', '102.0', '-3', '94.8', '101.5'])

number_of_samples = int(classroom_input("How many samples? "))

total_assay = 0.0
valid_count = 0
accepted_count = 0

for sample_no in range(1, number_of_samples + 1):
    assay_percent = float(classroom_input(f"Assay % for sample {sample_no}: "))

    if assay_percent < 0 or assay_percent > 200:
        print("Invalid classroom value skipped")
        continue

    total_assay += assay_percent
    valid_count += 1

    if 95.0 <= assay_percent <= 105.0:
        accepted_count += 1

if valid_count > 0:
    mean_assay = total_assay / valid_count
    print("Valid results:", valid_count)
    print("Inside classroom band:", accepted_count)
    print("Mean assay %:", round(mean_assay, 2))
else:
    print("No valid results were entered")
```

### Explanation

The for loop controls the planned repetitions. continue skips an invalid classroom entry before it reaches the accumulator. total_assay adds valid values, valid_count records how many valid values were processed, and accepted_count records how many meet the simulated band. The final if statement prevents division by zero.

### Interpretation

This program demonstrates the logic of processing repeated laboratory-style observations, separating invalid data from analyzable data, summarizing the valid values and counting values in a simulated classroom band.
