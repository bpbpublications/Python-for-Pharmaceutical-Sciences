## 07. Worked Example 3: Calculating Mean Assay Values from a Nested List

### Problem statement

Three simulated batches each have three assay readings. Calculate the arithmetic mean for each batch.

### Objective

Use a nested list with iteration, sum() and len() to summarise repeated measurements.

### Pharmacy context

Repeated assay measurements are often grouped by sample or batch. This example demonstrates a simple nested structure before introducing data-analysis libraries.

### Algorithm

1. Create a nested list with one inner list per batch.
2. Loop through each inner list.
3. Calculate sum(batch_readings) / len(batch_readings).
4. Print the batch number and mean.

### Variables and parameters

| Variable | Meaning |
| --- | --- |
| assay_batches | Nested list containing assay readings for each batch |
| batch_readings | Current inner list during iteration |
| mean_assay | Calculated arithmetic mean for one batch |

### Code and executed output

The output appears directly below the code cell.

```python
assay_batches = [
    [98.6, 99.0, 98.8],
    [100.1, 99.7, 100.0],
    [97.9, 98.2, 98.4]
]

for batch_number, batch_readings in enumerate(assay_batches, start=1):
    mean_assay = sum(batch_readings) / len(batch_readings)
    print(f'Batch {batch_number}: {mean_assay:.2f}%')
```

### Explanation

Each inner list is treated as one batch. enumerate(..., start=1) supplies a human-friendly batch number. The mean is formatted to two decimal places.

### Interpretation

The means are descriptive summaries of simulated assay data. Acceptance criteria must come from the relevant specification or validated method and are not inferred by this program.

### Objective

## 19. Worked Example 8: Medicine Inventory and QC Screening Record

### Problem statement

Create a small structured record for a medicine using multiple Python data structures, then display key information and unique QC observation labels.

### Objective

Integrate a list, tuple, dictionary and set in one beginner program.

### Pharmacy context

A medicine record can contain a fixed batch identity, several dissolution readings, labelled attributes and repeated observation categories. Different structures represent these meanings differently.

### Algorithm

1. Create a tuple for the fixed batch identity.
2. Create a list of dissolution readings.
3. Create a dictionary for labelled medicine attributes.
4. Create a list of repeated QC observation labels.
5. Convert the observation list to a set.
6. Print selected information.

### Variables and parameters

| Variable | Meaning |
| --- | --- |
| batch_identity | Tuple with batch code and dosage form |
| dissolution_percent | List of ordered dissolution readings |
| medicine_record | Dictionary containing labelled attributes |
| qc_observations | List of repeated QC labels |
| unique_observations | Set of unique QC labels |

### Code and executed output

The output appears directly below the code cell.

```python
batch_identity = ('B2401', 'Tablet')
dissolution_percent = [18.5, 42.0, 63.5, 78.2, 89.4, 96.1]
medicine_record = {
    'medicine_name': 'Paracetamol',
    'strength_mg': 500,
    'stock_units': 225,
    'batch_identity': batch_identity,
    'dissolution_percent': dissolution_percent
}
qc_observations = ['Label checked', 'Seal checked', 'Label checked', 'Appearance checked']
unique_observations = set(qc_observations)

print('Medicine:', medicine_record['medicine_name'])
print('Batch:', medicine_record['batch_identity'][0])
print('Final dissolution reading:', medicine_record['dissolution_percent'][-1], '%')
print('Unique QC observations:', unique_observations)
```

### Explanation

The dictionary is the main labelled record. It contains a tuple and a list as values. Nested indexing first selects a dictionary value and then selects an item from that structure. The set removes duplicate QC labels.

### Interpretation

The integrated example shows that Python structures are complementary rather than competing. The best structure depends on whether the data is ordered, fixed, labelled or required to be unique.
