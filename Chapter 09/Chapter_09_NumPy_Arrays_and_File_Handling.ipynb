## 13. Worked Program 1: Tablet Weight Array and Deviation from the Mean

### Problem statement

Store ten simulated tablet weights, calculate their mean, and calculate each tablet's deviation from the mean.

### Objective

To practise array creation, vectorized subtraction, and simple NumPy summary functions using tablet-weight data.

### Pharmacy context

The primary parameters are the unit (mg), the number of observations (10), and the reference used for deviation (the sample mean of 500.0 mg). This example describes variability around the sample mean only. It does not apply a pharmacopeial weight-variation acceptance rule.

### Algorithm

1. Import NumPy.
2. Create an array containing tablet weights in milligrams.
3. Calculate the mean weight with np.mean().
4. Subtract the mean from the full array to obtain deviations.
5. Display the mean and deviations.

### Variables and parameters

| Variable / Parameter | Meaning | Example / Unit |
| --- | --- | --- |
| tablet_weight_mg | Array of measured tablet weights | mg |
| mean_weight_mg | Arithmetic mean of all weights | mg |
| deviation_mg | Each weight minus the mean | mg |
| np.mean | Calculates arithmetic mean | function |

### Code and executed output

The output appears directly below the code cell.

````python
import numpy as np

tablet_weight_mg = np.array([498, 502, 501, 499, 503,
                             497, 500, 504, 496, 500])

mean_weight_mg = np.mean(tablet_weight_mg)
deviation_mg = tablet_weight_mg - mean_weight_mg

print("Mean weight (mg):", mean_weight_mg)
print("Deviation from mean (mg):", deviation_mg)
````

### Explanation

np.mean() reduces the array to one average value. Subtracting that scalar from the array is vectorized; NumPy performs the subtraction for every tablet. Positive values indicate weights above the calculated mean and negative values indicate weights below it.

### Interpretation

The primary parameters are the unit (mg), the number of observations (10), and the reference used for deviation (the sample mean of 500.0 mg). This example describes variability around the sample mean only. It does not apply a pharmacopeial weight-variation acceptance rule.

## 14. Worked Program 2: Assay Percentage from Measured Amount and Label Claim

### Problem statement

Calculate assay percentage for five simulated samples using measured amount and a common label claim.

### Objective

To demonstrate scalar broadcasting and rounding in a pharmaceutical calculation.

### Pharmacy context

label_claim_mg is the critical reference parameter. If its value or unit is wrong, every calculated percentage is wrong even though the code executes correctly. The values and interpretation here are classroom examples and are not a regulatory assay decision.

### Algorithm

1. Create an array of measured amounts.
2. Store the label claim as a scalar.
3. Calculate measured / label claim x 100 for every sample.
4. Round the displayed percentages to one decimal place.
5. Calculate the mean assay percentage.

### Variables and parameters

| Variable / Parameter | Meaning | Example / Unit |
| --- | --- | --- |
| measured_mg | Measured amount for each sample | mg |
| label_claim_mg | Reference amount used in calculation | 500 mg |
| assay_percent | Calculated percentage for each sample | % |
| decimals | Rounding precision used by np.round | 1 decimal |

### Code and executed output

The output appears directly below the code cell.

````python
import numpy as np

measured_mg = np.array([498, 505, 501, 497, 503])
label_claim_mg = 500

assay_percent = measured_mg / label_claim_mg * 100
mean_assay_percent = np.mean(assay_percent)

print("Assay (%):", np.round(assay_percent, 1))
print("Mean assay (%):", round(mean_assay_percent, 2))
````

### Explanation

The scalar 500 is used in the calculation for every element of measured_mg. NumPy broadcasting makes the expression compact and readable. The mean is calculated from the unrounded values and then displayed to two decimal places.

### Interpretation

label_claim_mg is the critical reference parameter. If its value or unit is wrong, every calculated percentage is wrong even though the code executes correctly. The values and interpretation here are classroom examples and are not a regulatory assay decision.

## 29. Worked Program 7: Dissolution Data Analysis and Report Generation

### Problem statement

Create an executable Jupyter workflow that imports dissolution_profile.csv, calculates a time-wise mean dissolution profile, evaluates the minimum release at 60 minutes against a clearly labelled simulated classroom threshold, and saves a summary CSV file.

### Objective

To integrate NumPy array handling and file handling in one pharmaceutical classroom problem.

### Pharmacy context

The 85% threshold is a teaching parameter, not a pharmacopeial or product-specific rule. Changing this parameter changes the decision without changing the measurements. The minimum 60-minute value is 93%, and the mean is 95%. Students should therefore report both the measured outputs and the assumption used for comparison. In real pharmaceutical work, acceptance criteria must come from the applicable approved specification or validated method.

### Algorithm

1. Create a Path to dissolution_profile.csv.
2. Set a simulated classroom threshold of 85%.
3. State the classroom decision condition: final_min_percent >= classroom_threshold_percent.
4. Load the numeric CSV data while skipping the header row.
5. Separate the time column and the six unit-release columns using slicing.
6. Calculate the row-wise mean release with axis=1.
7. Select the 60-minute row and calculate its minimum unit value.
8. Apply the stated classroom decision condition.
9. Create a two-column summary array and save it as a new CSV file.
10. Display the calculated values and the classroom result.

### Variables and parameters

| Variable / Parameter | Meaning | Example / Unit |
| --- | --- | --- |
| data_path | Input CSV location | dissolution_profile.csv |
| classroom_threshold_percent | Simulated teaching comparison | 85% |
| time_min | Sampling times | min |
| release_percent | Six unit profiles | % |
| mean_release_percent | Row-wise mean | % |
| final_min_percent | Minimum among six units at 60 min | % |
| status | Classroom comparison result | MEETS / DOES NOT MEET |
| axis=1 | Direction of mean calculation | across units |

### Code and executed output

````python
from pathlib import Path
import numpy as np

data_path = (DATA_DIR / "dissolution_profile.csv")
classroom_threshold_percent = 85.0  # simulated teaching value

if not data_path.exists():
    raise FileNotFoundError(
        "Run the Chapter 9 classroom dataset setup cell in Section 9.12.1 first."
    )

data = np.loadtxt(data_path, delimiter=",", skiprows=1)
time_min = data[:, 0]
release_percent = data[:, 1:]

mean_release_percent = np.mean(release_percent, axis=1)
final_min_percent = np.min(release_percent[-1, :])

if final_min_percent >= classroom_threshold_percent:
    status = "MEETS CLASSROOM THRESHOLD"
else:
    status = "DOES NOT MEET CLASSROOM THRESHOLD"

summary = np.column_stack((time_min, mean_release_percent))
np.savetxt(OUTPUT_DIR / "dissolution_summary.csv",
           summary,
           delimiter=",",
           fmt=["%.0f", "%.2f"],
           header="time_min,mean_release_percent",
           comments="")

print("Mean at 60 min (%):", round(mean_release_percent[-1], 2))
print("Minimum unit at 60 min (%):", final_min_percent)
print("Classroom threshold (%):", classroom_threshold_percent)
print("Result:", status)
````

### Explanation

The program keeps the source data in the input file and writes calculated means to a separate summary file. release_percent[-1, :] selects every unit in the final row.
