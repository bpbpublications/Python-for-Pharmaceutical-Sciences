## 08. Worked Program 1: Create and Interpret a Labelled Dose Series

### Problem statement

Create a Pandas Series containing four simulated dose values indexed by subject ID. Display the Series, calculate the mean and maximum, and retrieve the value for subject P02.

### Objective

To practise creation and basic interpretation of a labelled one-dimensional pharmaceutical variable in Pandas.

### Pharmacy context

data contains four classroom values. index connects each value to a subject label. name="dose_mg" records the unit in the Series name. mean() returns 512.5 mg for these four simulated observations, max() returns 650 mg, and loc["P02"] shows that the 650 mg observation belongs to P02. These outputs describe the dataset only; they are not dose recommendations.

### Algorithm

1. Import Pandas as pd.
2. Create a Series containing four dose values.
3. Use subject identifiers as the index.
4. Assign the name dose_mg to preserve the unit in the variable label.
5. Calculate the mean and maximum dose values.
6. Retrieve the dose for P02 using label-based access.
7. Display the results with units.

### Variables and parameters

| Variable / Parameter | Meaning | Example / Unit |
| --- | --- | --- |
| dose_mg | Series of simulated doses | mg |
| index | Subject labels | P01-P04 |
| name | Name of the Series | dose_mg |
| mean() | Arithmetic mean of the Series | mg |
| max() | Largest value in the Series | mg |
| loc["P02"] | Label-based lookup | P02 |

### Code and executed output

The output appears directly below the code cell.

````python
import pandas as pd

dose_mg = pd.Series(
    [500, 650, 400, 500],
    index=["P01", "P02", "P03", "P04"],
    name="dose_mg"
)

print(dose_mg)
print("Mean dose:", round(dose_mg.mean(), 1), "mg")
print("Maximum dose:", dose_mg.max(), "mg")
print("Subject P02 dose:", dose_mg.loc["P02"], "mg")
````
