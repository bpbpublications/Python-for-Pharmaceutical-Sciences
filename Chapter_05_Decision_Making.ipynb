The 0-14 domain and the simplified classification are used for programming education. A valid numeric entry is not automatically a valid pharmaceutical measurement or specification result.

### Algorithm

1. Receive the input as text and strip surrounding whitespace.
2. Attempt float conversion inside try.
3. If conversion succeeds, test 0 <= ph_value <= 14.

5. If conversion fails, catch ValueError and display a numeric-input message.

### Variables and parameters

| Variable | Example | Purpose |
| --- | --- | --- |
| raw_value | "6.8" | Text returned by input() |
| ph_value | 6.8 | Converted numeric value when conversion succeeds |

### Important notes

Saved output uses these sample entries in order: `6.8`. Set INTERACTIVE_INPUT = True in the setup cell to type your own entries.

### Code and executed output

The output appears directly below the code cell.

```python
_sample_answers = iter(['6.8'])

raw_value = classroom_input("Enter pH value: ").strip()

try:
    ph_value = float(raw_value)

    if 0 <= ph_value <= 14:
        print("Value accepted for this classroom exercise.")
    else:
        print("Value is outside the exercise domain 0 to 14.")

except ValueError:
    print("Please enter a numeric value.")
```

### Code and executed output

The output appears directly below the code cell.

```python
from datetime import date

# Simulated single-record data
batch_code = "PCM-B17"
current_stock = 7
reorder_level = 10
reference_date = date(2026, 8, 21)
expiry_date = date(2026, 12, 31)
temperature_c = 23
humidity_percent = 65
seal_ok = True
label_ok = False

# 1. Batch-code validation
batch_missing = batch_code.strip() == ""
if batch_missing:
    print("Batch: MISSING")
else:
    print("Batch:", batch_code.strip())

# 2. Stock decision
stock_low = current_stock < reorder_level
if stock_low:
    print("Stock: REORDER in classroom example")
else:
    print("Stock: SUFFICIENT in classroom example")

# 3. Expiry-date relation
if expiry_date < reference_date:
    print("Expiry: before reference date")
elif expiry_date == reference_date:
    print("Expiry: equals reference date")
else:
    print("Expiry: after reference date")

# 4. Nested simulated storage check
if 20 <= temperature_c <= 25:
    if 40 <= humidity_percent <= 60:
        print("Storage: both classroom conditions in range")
    else:
        print("Storage: humidity outside classroom range")
else:
    print("Storage: temperature outside classroom range")

# 5. Simulated packaging check
packaging_ok = seal_ok and label_ok
if packaging_ok:
    print("Packaging: both simulated checks passed")
else:
    print("Packaging: at least one simulated check needs attention")

# 6. Overall teaching attention flag
packaging_issue = not packaging_ok
needs_attention = stock_low or batch_missing or packaging_issue

if needs_attention:
    print("Overall: record needs attention under the classroom rules")
else:
    print("Overall: no attention flag from the classroom rules")
```

### Explanation

The batch code is present. Stock is below the classroom reorder level. The expiry date is after the fixed reference date. Temperature is inside the simulated range, so humidity is checked and found outside its simulated range. The seal check is True but the label check is False, so packaging_ok is False. Because stock_low or packaging_issue is True, the final attention flag is True.
