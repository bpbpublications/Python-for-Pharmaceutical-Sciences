## 04. 4.3.1 Program 4.4: Slicing a structured batch code

### Problem statement

Extract `PCM`, `500` and `B17` from the simulated code `PCM-500-B17` by using slicing.

### Objective

To apply start and stop positions in string slices and understand the excluded stop index.

### Pharmacy context

A batch identifier may follow a documented internal format. This example assumes a fixed classroom structure only; real batch-code formats depend on organisational systems and should not be guessed from appearance.

### Algorithm

1. Store the batch code.
2. Slice from the beginning to index 3 to obtain the prefix.
3. Slice indexes 4 to 6 by stopping before index 7.
4. Slice from index 8 to the end.
5. Print the three components.

### Variables and parameters

| Variable | Meaning | Type/Unit |
| --- | --- | --- |
| batch_code | Structured simulated identifier | str |

### Code and executed output

The output appears directly below the code cell.

````python
batch_code = "PCM-500-B17"

print(batch_code[:3])
print(batch_code[4:7])
print(batch_code[8:])
````

### Explanation

`[:3]` starts at the beginning and stops before index 3. `[4:7]` selects indexes 4, 5 and 6. `[8:]` starts at index 8 and continues through the final character.

### Interpretation

Slicing can extract fields from a truly fixed-format identifier. If a reliable delimiter exists, `split()` is often clearer and less fragile than fixed character positions.

## 08. 4.6.4.1 Program 4.8: Searching a simulated pharmaceutical record

### Problem statement

Find the word `Tablet`, count field separators and test the beginning and ending of a delimited text record.

### Objective

To apply `find()`, `count()`, `startswith()` and `endswith()` to structured text.

### Pharmacy context

Delimited records store several fields in one text line by placing a known separator between values. This classroom record uses the vertical bar `|`; real data files should follow a documented format.

### Algorithm

1. Store the structured text record.
2. Use `find()` to locate `Tablet`.
3. Count the vertical-bar separators.
4. Check the beginning and ending text.
5. Display the results.

### Variables and parameters

| Variable | Meaning | Type/Unit |
| --- | --- | --- |
| record | Simulated delimited record | str |

### Code and executed output

The output appears directly below the code cell.

````python
record = "Paracetamol|500 mg|Tablet|Batch B17"

print(record.find("Tablet"))
print(record.count("|"))
print(record.startswith("Paracetamol"))
print(record.endswith("B17"))
````

### Explanation

### Interpretation

These methods can screen the form of a simulated record. A positive text check confirms only the presence or position of characters; it does not establish pharmaceutical validity.

## 15. 4.10.2 Program 4.15: `statistics`: Summarizing simulated tablet weights

### Problem statement

Calculate the mean, median and sample standard deviation of five simulated tablet weights.

### Objective

To use selected functions from the standard library `statistics` module.

### Pharmacy context

Tablet weight is a measurable property used in pharmaceutical manufacturing and quality work. The values here are simulated solely to demonstrate descriptive statistics; no acceptance specification is applied.

### Algorithm

1. Import `mean`, `median` and `stdev`.
2. Store five simulated weights.
3. Calculate the three descriptive statistics.
4. Display the results, rounding the standard deviation for readability.

### Variables and parameters

| Variable | Meaning | Type/Unit |
| --- | --- | --- |
| weights_mg | Five simulated tablet weights | list of int/mg |
| average_weight | Arithmetic mean | int or float/mg |
| median_weight | Median | int or float/mg |
| sample_sd | Sample standard deviation | float/mg |

### Code and executed output

The output appears directly below the code cell.

````python
from statistics import mean, median, stdev

weights_mg = [498, 501, 500, 502, 499]
average_weight = mean(weights_mg)
median_weight = median(weights_mg)
sample_sd = stdev(weights_mg)

print(average_weight)
print(median_weight)
print(round(sample_sd, 2))
````

### Explanation

mean() calculates the arithmetic average, median() gives the middle ordered value and stdev() calculates the sample standard deviation. Here, round(sample_sd, 2) returns a rounded value for printing; sample_sd itself is unchanged.

### Interpretation

These statistics summarise the simulated measurements. Whether any real tablet weights are acceptable requires the appropriate validated pharmaceutical method and specification.

## 29. 4.17.1 Program 4.24: Integrated medicine record and output path

### Problem statement

A simulated record contains a medicine name, strength and quantity separated by vertical bars. Clean the text, extract the values, convert numeric text to integers, calculate how many whole packs of 10 units are required, and build the path for a cleaned summary file.

### Objective

To integrate string methods, type conversion, `math.ceil()` and `pathlib.Path` without using conditional statements or loops.

### Pharmacy context

The record represents classroom inventory-style text. Pack calculations are used only to demonstrate programming and must not be treated as dispensing, procurement or stock-control instructions for real medicines.

### Algorithm

1. Import `ceil` and `Path`.
2. Store the raw delimited record.
3. Remove surrounding whitespace.
4. Split at the vertical bars and strip each known field without a loop.
5. Convert strength and quantity text to integers.
6. Calculate the number of whole packs using `ceil()`.
7. Build a relative output path in the `outputs` folder.
8. Print the cleaned fields, calculation and path.

### Variables and parameters

| Variable | Meaning | Type/Unit |
| --- | --- | --- |
| raw_record | Original delimited text | str |
| raw_parts | Three split text fields | list of str |
| strength_mg | Converted strength value | int/mg |
| quantity | Simulated quantity | int/units |
| packs_required | Whole-pack calculation | int/packs |
| output_path | Relative output location | Path |

### Code and executed output

The output appears directly below the code cell.

````python
from math import ceil
from pathlib import Path

raw_record = " paracetamol | 500 | 73 "
clean_record = raw_record.strip()
raw_parts = clean_record.split("|")

medicine_name = raw_parts[0].strip().title()
strength_mg = int(raw_parts[1].strip())
quantity = int(raw_parts[2].strip())

pack_size = 10
packs_required = ceil(quantity/pack_size)
output_path = Path("outputs")/"clean_inventory_summary.txt"

print(medicine_name)
print(strength_mg)
print(quantity)
print(packs_required)
print(output_path)
````

### Explanation

The record is trimmed and split into three fields. Each field is stripped explicitly. int() converts numeric text, ceil() produces a whole-pack count, and Path constructs the intended output location. This program prints the results; it does not write the summary file.

### Interpretation

The example demonstrates a transparent data-preparation workflow in Jupyter. It assumes the input structure is correct; checking missing, invalid or boundary values is introduced in Chapter 5 with conditional statements.
