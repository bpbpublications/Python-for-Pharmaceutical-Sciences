## 08. 3.1.8.1 Pharmacy example: Operator precedence in percentage strength calculation

### Problem statement

A student needs to calculate the percentage strength of a solution using drug mass and solution volume, while ensuring that the expression is evaluated in the correct order.

### Objective

To demonstrate the importance of parentheses in percentage strength calculations and to avoid confusion caused by operator precedence.

### Pharmacy context

The output shows a strength of 0.25% w/v because 250 mg is first converted to 0.25 g, then divided by 100 mL, and finally multiplied by 100. From a pharmacy application point of view, this reinforces that unit conversion and correct order of operations are essential before interpreting any calculated formulation value.

### Algorithm

1. Store `iter(['250', '100'])` in `_sample_answers`.
2. Store `float(classroom_input('Enter drug mass in mg: '))` in `drug_mass_mg`.
3. Store `float(classroom_input('Enter solution volume in mL: '))` in `solution_volume_ml`.
4. Store `drug_mass_mg / 1000` in `drug_mass_g`.
5. Store `drug_mass_g / solution_volume_ml * 100` in `strength_percent`.
6. Display the results for inspection.

### Important notes

Saved output uses these sample entries in order: `250`, `100`. Set INTERACTIVE_INPUT = True in the setup cell to type your own entries.

### Code and executed output

The output appears directly below the code cell.

```python
_sample_answers = iter(['250', '100'])

drug_mass_mg = float(classroom_input("Enter drug mass in mg: "))
solution_volume_ml = float(classroom_input("Enter solution volume in mL: "))
drug_mass_g = drug_mass_mg / 1000
strength_percent = (drug_mass_g / solution_volume_ml) * 100
print("Strength:", strength_percent, "% w/v")
```

### Explanation

The mass is first converted from milligrams to grams. Division and multiplication have equal precedence and group from left to right. The parentheses make the intended percentage-strength calculation explicit; the result is unchanged without them in this expression.

### Interpretation

The output shows a strength of 0.25% w/v because 250 mg is first converted to 0.25 g, then divided by 100 mL, and finally multiplied by 100. From a pharmacy application point of view, this reinforces that unit conversion and correct order of operations are essential before interpreting any calculated formulation value.

## Operator precedence: verified examples

**Problem:** Predict calculations that combine powers, unary signs, multiplication, addition, comparisons and Boolean operations.

**Pharmacy context:** Parentheses make classroom formulation and stock calculations easier to audit.

**Algorithm:** Evaluate explicit parentheses first; apply Python precedence; compare the output with the predicted value. Power is right-associative. Unary minus to the left of a power is applied after that power; a signed exponent is allowed on the right.

Highest to lowest among these operators: `**`; unary `+ - ~`; `* / // % @`; `+ -`; `<< >>`; `&`; `^`; `|`; comparisons (including membership and identity); `not`; `and`; `or`. Parentheses control grouping. Assignment statements are not a level in this expression-precedence list.

```python
print("2 + 3 * 4 =", 2 + 3 * 4)
print("(2 + 3) * 4 =", (2 + 3) * 4)
print("2 ** 3 ** 2 =", 2 ** 3 ** 2)
print("-2 ** 2 =", -2 ** 2)
print("(-2) ** 2 =", (-2) ** 2)
print("2 ** -2 =", 2 ** -2)
print("2 <= 6 <= 8 =", 2 <= 6 <= 8)
print("not False and True or False =", not False and True or False)
```

**Explanation:** The results are 14, 20, 512, −4, 4, 0.25, True and True. Powers associate to the right. Comparisons can chain; `not` binds more tightly than `and`, which binds more tightly than `or`.
