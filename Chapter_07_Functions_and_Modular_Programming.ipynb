## 07. 7.8.1 Returning More Than One Value

### Problem statement

Demonstrate returning More Than One Value using the simulated values or objects in the cell below.

### Objective

Understand how the Python operations in this example support returning more than one value and interpret the result correctly.

### Pharmacy context

Reusable functions make assumptions, inputs and units explicit. Concentration, yield, BMI and dose-related calculations are educational examples and must not be used to make individual clinical decisions.

### Algorithm

1. Define `batch_summary(accepted_units, rejected_units)` with the calculation or action shown in its body.
2. Store `batch_summary(980, 20)` in `(total, rejection)`.
3. Display the results for inspection.

### Important notes

Input conditions: use finite numerical values, positive denominators and compatible units. These introductory functions assume valid inputs; a successful calculation does not validate a clinical or QC decision.

### Code and executed output

The output appears directly below the code cell.

```python
def batch_summary(accepted_units, rejected_units):
    total_units = accepted_units + rejected_units
    rejection_percent = (rejected_units / total_units) * 100
    return total_units, rejection_percent

total, rejection = batch_summary(980, 20)
print("Total units:", total)
print("Rejection %:", round(rejection, 2))
```

### Explanation

Python can return more than one value by separating the values with commas. The caller can then unpack the returned values into separate variables. This is useful when one calculation naturally produces related results.

### Interpretation

| Variable | Meaning |
| --- | --- |
| required_dose_mg | Given dose value in milligrams. |

## 25. 7.22 Integrated Mini-Project: Reusable Pharmacy Calculation Toolkit

### Problem statement

Create a simple pharmacy calculation toolkit containing reusable functions for concentration, percentage yield, total tablet stock, and stock status. Use simulated classroom data and display the final results clearly.

### Objective

To integrate function definition, parameters, arguments, return values, conditional logic, and modular design in one beginner-level B.Pharm program.

### Pharmacy context

The mini-project demonstrates how reusable functions can support pharmaceutical calculations and inventory exercises. The values and thresholds are simulated classroom data; real laboratory, manufacturing, inventory, or clinical decisions must use validated procedures and approved criteria.

### Algorithm

1. Define calculate_concentration().
2. Define calculate_percent_yield().
3. Define calculate_total_tablets().
4. Define stock_status().
5. Call each function with simulated data.
6. Store returned values.
7. Display a short summary with units and status.

### Variables and parameters

| Variable | Meaning |
| --- | --- |
| amount_mg | Amount of substance in mg. |
| volume_ml | Volume in mL. |
| actual_yield_g | Actual yield in g. |
| theoretical_yield_g | Theoretical yield in g. |
| strips | Number of strips. |
| tablets_per_strip | Tablets in each strip. |
| available_stock | Current stock units. |
| reorder_level | Classroom reorder threshold. |

### Important notes

Input conditions: use finite numerical values, positive denominators and compatible units. These introductory functions assume valid inputs; a successful calculation does not validate a clinical or QC decision.

### Code and executed output

The output appears directly below the code cell.

```python
def calculate_concentration(amount_mg, volume_ml):
    return amount_mg / volume_ml

def calculate_percent_yield(actual_yield_g, theoretical_yield_g):
    return (actual_yield_g / theoretical_yield_g) * 100

def calculate_total_tablets(strips, tablets_per_strip):
    return strips * tablets_per_strip

def stock_status(available_stock, reorder_level):
    if available_stock < reorder_level:
        return "REORDER"
    return "SUFFICIENT"

# Simulated classroom data
concentration = calculate_concentration(300, 60)
yield_percent = calculate_percent_yield(96, 100)
total_tablets = calculate_total_tablets(15, 10)
status = stock_status(total_tablets, 200)

print("PHARMACY TOOLKIT SUMMARY")
print("Concentration:", round(concentration, 2), "mg/mL")
print("Percentage yield:", round(yield_percent, 2), "%")
print("Total tablets:", total_tablets)
print("Stock status:", status)
```

### Explanation

The program is divided into four functions. Each function has a single responsibility and returns a result. The main part of the program calls these functions, stores the returned values, and prints the final summary. This structure is easier to test and extend than one long block of repeated calculations.

### Interpretation

The mini-project demonstrates how reusable functions can support pharmaceutical calculations and inventory exercises. The values and thresholds are simulated classroom data; real laboratory, manufacturing, inventory, or clinical decisions must use validated procedures and approved criteria.
